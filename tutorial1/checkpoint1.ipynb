<h2> Feature Engineering </h2>

After exploring the data, you can choose to modify features that you would use to predict the performance of the students on your chosen response variable. 

You can also create your own features. For example, perhaps you figured that maybe a state's expenditure per student may affect their overall academic performance so you create a expenditure_per_student feature.

Use this space to modify or create features.

```python
# @@@ 5
# Normalize spending for population size; zero enrollment cannot be a divisor.
enrollment = df["GRADES_ALL_G"].fillna(df["ENROLL"]).replace(0, np.nan)
df["EXPENDITURE_PER_STUDENT"] = df["TOTAL_EXPENDITURE"] / enrollment
# Use a contemporaneous reading score and spending to explain math performance.
features = ["AVG_READING_4_SCORE", "EXPENDITURE_PER_STUDENT", "YEAR"]
```

Reading scores summarize related academic performance; spending per student adjusts expenditure for enrollment; year captures broad time trends. Median imputation retains rows with missing predictors, and scaling makes Ridge regularization comparable across features. This is a contemporaneous association model, not a forecast or a causal estimate of spending effects.

<h2> Data Creation </h2>

_Use this space to create train/test data_

```python
from sklearn.model_selection import train_test_split
```

```python
# @@@ 8
# Keep observed targets only; impute predictors later using training data only.
model_data = df.dropna(subset=[target]).copy()
X = model_data[features]
y = model_data[target]
```

```python
# @@@ 9 
# Hold out entire states to avoid sharing repeated observations of a state.
from sklearn.model_selection import GroupShuffleSplit
splitter = GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=42)
train_idx, test_idx = next(splitter.split(X, y, groups=model_data["STATE"]))
X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]
assert set(model_data.iloc[train_idx]["STATE"]).isdisjoint(model_data.iloc[test_idx]["STATE"])
print("Train/test rows:", len(X_train), len(X_test))
```

<h2> Prediction </h2>

ML Models [Resource](https://medium.com/@vijaya.beeravalli/comparison-of-machine-learning-classification-models-for-credit-card-default-data-c3cf805c9a5a)

```python
# @@@ 10
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge
from sklearn.dummy import DummyRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
# @@@ 11
# Median imputation and scaling are learned inside fit on training rows only.
model = make_pipeline(SimpleImputer(strategy="median"), StandardScaler(), Ridge(alpha=1.0))
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

## Evaluation

Choose some metrics to evaluate the performance of your model, some of them are mentioned in the tutorial.

```python
# @@@ 12
baseline = DummyRegressor(strategy="mean").fit(X_train, y_train)
results = []
for name, fitted in [("Mean baseline", baseline), ("Ridge", model)]:
    pred = fitted.predict(X_test)
    results.append({"model": name, "MAE": mean_absolute_error(y_test, pred),
                    "RMSE": np.sqrt(mean_squared_error(y_test, pred)), "R2": r2_score(y_test, pred)})
display(pd.DataFrame(results).set_index("model"))
```

We have copied over the graphs that visualize the model's performance on the training and testing set. 

Change `col_name` and modify the call to `plt.ylabel()` to isolate how a single predictor affects the model.

```python
# @@@ 13
col_name = "AVG_READING_4_SCORE"
plt.figure(figsize=(9, 5))
plt.scatter(X_train[col_name], y_train, alpha=0.5, color="red", label="Observed")
plt.scatter(X_train[col_name], model.predict(X_train), alpha=0.5, color="black", label="Predicted")
plt.xlabel("Average grade 4 reading score")
plt.ylabel("Average grade 4 math score")
plt.title("Model behavior on the train set")
plt.legend()
plt.show()
```

```python
# @@@ 14
col_name = "AVG_READING_4_SCORE"
plt.figure(figsize=(9, 5))
plt.scatter(X_test[col_name], y_test, alpha=0.5, color="blue", label="Observed")
plt.scatter(X_test[col_name], model.predict(X_test), alpha=0.5, color="black", label="Predicted")
plt.xlabel("Average grade 4 reading score")
plt.ylabel("Average grade 4 math score")
plt.title("Model behavior on the test set")
plt.legend()
plt.show()
```

The held-out-state metrics estimate performance on states excluded from training. Compare Ridge with the mean baseline above; R² can be negative for a poor model. The data are aggregated state averages, so these results do not predict individual student outcomes. Missingness, changing tests, and unmeasured state differences limit interpretation.


**Observed result:** Ridge achieved test MAE 3.73, RMSE 4.76, and R² 0.712, versus MAE 6.81 and RMSE 8.88 for the training-mean baseline. It captures substantial variation across held-out states, but prediction errors of several score points remain.
